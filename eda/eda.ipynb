### Split Data

```python
#split the dataset into reference(80%) and production(20%) sets
reference, production = train_test_split(
    df,
    test_size=0.2,
    stratify=df["TARGET"],  
    random_state=42,        
)

reference.shape, production.shape
```

```python
#stratify evidence
print(reference["TARGET"].value_counts(normalize=True))
print(production["TARGET"].value_counts(normalize=True))
```

```python
# Salve the reference and production datasets to CSV files
reference.to_csv("../data/reference.csv", index=False)
production.to_csv("../data/production.csv", index=False)
```
